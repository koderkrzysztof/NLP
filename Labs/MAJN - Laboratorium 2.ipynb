```python
# Zadanie 5
# Dla dokumentu II oraz dwóch wybranych dokumentów ze zbioru III utwórz reprezentację 1 z n
# dla tokenów składających się z pojedynczych słów. Następnie spróbuj utworzyć tę
# reprezentację uwzględniając: 2-gramy, 3-gramy, stop listę, stemming.

import re
from functools import lru_cache

import nltk
import numpy as np
import pandas as pd
from nltk.stem.porter import PorterStemmer
from nltk.util import ngrams
from scipy import sparse

# tokenizator: TweetTokenizer, małe litery przez str.lower()
tokenizer = TweetTokenizer()

# stop lista z NLTK
nltk.download('stopwords')
stop_lista = set(nltk.corpus.stopwords.words('english'))
print("Rozmiar stop listy NLTK:", len(stop_lista))

# stemmer Portera z NLTK;
stemmer = PorterStemmer()


def tokeny(tekst, stop=False, stemming=False):
    """Małe litery + tokenizacja (+ stop lista i stemming)."""
    tokens = [t for t in tokenizer.tokenize(tekst.lower()) if re.search(r"\w", t)]
    if stop:
        tokens = [t for t in tokens if t not in stop_lista]
    if stemming:
        tokens = [stemmer.stem(t) for t in tokens]
    return tokens


def ngramy(tokens, zakres=(1, 1)):
    """Wszystkie n-gramy z zadanego zakresu (np. (1, 2) = pojedyncze słowa + 2-gramy)."""
    wynik = []
    for n in range(zakres[0], zakres[1] + 1):
        wynik += [" ".join(gram) for gram in ngrams(tokens, n)]
    return wynik


def zbuduj_slownik(listy_tokenow):
    """Słownik (z eliminacją duplikatów) """
    wszystkie = [t for tokens in listy_tokenow for t in tokens]
    return sorted(list(set(wszystkie)))


def jeden_z_n(tokens, vocabulary):
    """Macierz (liczba tokenów x rozmiar słownika)."""
    one_hot = np.zeros([len(tokens), len(vocabulary)])
    for i, word in enumerate(tokens):
        one_hot[i, vocabulary.index(word)] = 1
    return one_hot


dokumenty = {
    "II (sun.txt)": tokenizer_test_document,
    "III nr 0": newsgroups.data[0],
    "III nr 1": newsgroups.data[1],
}

# (nazwa, zakres n-gramów, stop lista, stemming)
konfiguracje = [
    ("1-gramy",                         (1, 1), False, False),
    ("1+2-gramy",                       (1, 2), False, False),
    ("1+2+3-gramy",                     (1, 3), False, False),
    ("1-gramy + stop lista",            (1, 1), True,  False),
    ("1-gramy + stemming",              (1, 1), False, True),
    ("1-gramy + stop lista + stemming", (1, 1), True,  True),
    ("1+2+3-gramy + stop + stemming",   (1, 3), True,  True),
]

# przykład: początek reprezentacji 1 z n dla pierwszego zdania dokumentu I (Roosevelt)
toks_r = tokeny(roosevelt_sentence)
slownik_r = zbuduj_slownik([toks_r])
print("Tokeny:", toks_r)
print("Słownik:", slownik_r)
# czytelniejsza wersja w postaci ramki danych Pandas
pd.DataFrame(jeden_z_n(toks_r, slownik_r), columns=slownik_r)
```

```python
# osobny słownik dla każdego dokumentu
wiersze = []
for nazwa_konf, zakres, stop, stemming in konfiguracje:
    for nazwa_dok, tekst in dokumenty.items():
        toks = ngramy(tokeny(tekst, stop, stemming), zakres)
        slownik = zbuduj_slownik([toks])
        macierz = jeden_z_n(toks, slownik)
        wiersze.append({
            "konfiguracja": nazwa_konf,
            "dokument": nazwa_dok,
            "tokeny n": len(toks),
            "rozmiar słownika": len(slownik),
            "kształt macierzy 1 z n": macierz.shape,
        })
tabela_osobne = pd.DataFrame(wiersze)
pd.set_option("display.width", 200)
print(tabela_osobne.to_string(index=False))
```

```python
# wpływ wyboru tokenizatora, stop listy i stemmera na rozmiar słownika
from nltk.stem import SnowballStemmer
from nltk.stem.lancaster import LancasterStemmer
from nltk.tokenize import TreebankWordTokenizer
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

treebank = TreebankWordTokenizer()
tokenizatory = {
    "str.split()": lambda t: t.split(),
    "Treebank":    lambda t: treebank.tokenize(t),
    "Tweet":       lambda t: tokenizer.tokenize(t),
}
stop_listy = {
    "NLTK":         stop_lista,
    "spaCy":        nlp.Defaults.stop_words,
    "scikit-learn": ENGLISH_STOP_WORDS,
}
stemmery = {
    "Porter":    PorterStemmer(),
    "Snowball":  SnowballStemmer("english"),
    "Lancaster": LancasterStemmer(),
}


def tokeny_wariant(tekst, tok, stop=None, stemmer=None):
    tokens = [t for t in tokenizatory[tok](tekst.lower()) if re.search(r"\w", t)]
    if stop is not None:
        tokens = [t for t in tokens if t not in stop]
    if stemmer is not None:
        tokens = [stemmer.stem(t) for t in tokens]
    return tokens


wiersze = []
# 1) tokenizator (bez stop listy i stemmingu) oraz tokenizator + stop lista NLTK
for tok in tokenizatory:
    for nazwa_dok, tekst in dokumenty.items():
        bez = tokeny_wariant(tekst, tok)
        ze = tokeny_wariant(tekst, tok, stop=stop_lista)
        wiersze.append({"tokenizator": tok, "dokument": nazwa_dok,
                        "tokeny": len(bez), "słownik": len(set(bez)),
                        "tokeny po stop liście NLTK": len(ze), "słownik po stop liście NLTK": len(set(ze))})
print(pd.DataFrame(wiersze).to_string(index=False))

# 2) wybór stop listy (tokenizator Tweet)
wiersze = []
for nazwa_stop, stop in stop_listy.items():
    for nazwa_dok, tekst in dokumenty.items():
        toks = tokeny_wariant(tekst, "Tweet", stop=stop)
        wiersze.append({"stop lista": nazwa_stop, "rozmiar listy": len(stop), "dokument": nazwa_dok,
                        "tokeny": len(toks), "słownik": len(set(toks))})
print()
print(pd.DataFrame(wiersze).to_string(index=False))

# 3) wybór stemmera (tokenizator Tweet, bez stop listy)
wiersze = []
for nazwa_stem, st in stemmery.items():
    for nazwa_dok, tekst in dokumenty.items():
        toks = tokeny_wariant(tekst, "Tweet", stemmer=st)
        wiersze.append({"stemmer": nazwa_stem, "dokument": nazwa_dok, "słownik": len(set(toks))})
print()
print(pd.DataFrame(wiersze).to_string(index=False))
```

**Odpowiedzi do Zadania 5**

**Jaki tokenizator i stop listę wybrałem i dlaczego?**
- **Tokenizator:** `TweetTokenizer` z usunięciem tokenów składających się wyłącznie ze znaków interpunkcyjnych. Odcina kropki i przecinki od słów w całym tekście, zostawia adresy e-mail i emotikony, a małe litery łączą `The` i `the` w jeden wpis.
- **Stop lista:** angielska lista z NLTK (`nltk.corpus.stopwords`, 198 słów). Słowa funkcyjne (`the`, `of`, `is`) dominują w tekście i niewiele mówią o treści.

**Jaki ten wybór miał wpływ na rozmiar słownika i strukturę reprezentacji 1 z n?**
- **Tokenizator:** nie zmienia struktury reprezentacji, a rozmiar słownika tylko nieznacznie. Zmienia głównie jakość tokenów: `str.split()` zostawia interpunkcję przyklejoną do słów (`system.` obok `system`), Treebank rozbija adresy e-mail, a Tweet radzi sobie z nimi najlepiej.
- **Stop lista:** usuwa najczęstsze słowa, więc kurczą się **oba wymiary** macierzy: maleje liczba wierszy (tokenów) i kolumn (słownika). Im dłuższa lista, tym mniejsza macierz.

**Jaki wpływ na rozmiar słownika i strukturę reprezentacji miało dodanie 2-gramów i 3-gramów oraz wykorzystanie stemmera i stop listy?**
- **2-gramy i 3-gramy:** liczba tokenów rośnie, słownik również rośnie, bo większość n-gramów występuje tylko raz. Macierz rośnie wielokrotnie.
- **Stemming:** nie zmienia liczby tokenów, tylko trochę zmniejsza słownik, bo łączy różne formy tego samego słowa (`field` i `fields`). Wybór stemmera (Porter, Snowball, Lancaster) ma niewielki wpływ.
- **Stop lista + stemming:** oba zabiegi zmniejszają macierz.
- Stop lista jest usuwana **przed** budową n-gramów, więc powstają n-gramy łączące słowa, które w tekście nie sąsiadowały.


```python
# wspólny słownik dla wszystkich 3 dokumentów
wiersze = []
for nazwa_konf, zakres, stop, stemming in konfiguracje:
    tokeny_dok = {nazwa: ngramy(tokeny(t, stop, stemming), zakres) for nazwa, t in dokumenty.items()}
    wspolny = zbuduj_slownik(tokeny_dok.values())
    macierze = {nazwa: jeden_z_n(t, wspolny) for nazwa, t in tokeny_dok.items()}
    wiersze.append({
        "konfiguracja": nazwa_konf,
        "wspólny słownik": len(wspolny),
        "suma słowników osobnych": sum(len(zbuduj_slownik([t])) for t in tokeny_dok.values()),
        "kształty macierzy": [m.shape for m in macierze.values()],
    })
tabela_wspolny = pd.DataFrame(wiersze)
print(tabela_wspolny.to_string(index=False))

# macierz gęsta a macierz rzadka (CSR) dla dokumentu II, 1-gramy
toks = ngramy(tokeny(tokenizer_test_document))
macierz = jeden_z_n(toks, zbuduj_slownik([toks]))
rzadka = sparse.csr_matrix(macierz)
print("Gęsta:", macierz.nbytes, "B | rzadka CSR:", rzadka.data.nbytes + rzadka.indices.nbytes + rzadka.indptr.nbytes, "B")
print("Lista numerów słów (same indeksy):", np.array(rzadka.indices).nbytes, "B")
```

**Jak zwiększy się słownik i zmieni reprezentacja 1 z n, jeżeli słownik będzie wspólny dla wszystkich 3 dokumentów?**
- Wspólny słownik to **suma zbiorów** słów z trzech dokumentów. Jest mniejszy niż suma słowników osobnych (370 < 415), bo część słów powtarza się (`the`, `of`, `is`), ale większy niż słownik każdego dokumentu z osobna (230, 89 i 96).
- Każda z trzech macierzy ma teraz tyle samo kolumn (445 x 370, 121 x 370, 127 x 370). Ta sama kolumna oznacza to samo słowo we wszystkich dokumentach, więc reprezentacje są porównywalne. Kosztem jest dużo kolumn zawierających same zera, więc razem macierze zajmują 250 kB zamiast 122 kB.
- Dla 2-gramów i 3-gramów prawie nie ma oszczędności ze wspólnego słownika (1639 vs1696), bo n-gramy rzadko powtarzają się między różnymi tekstami.
- Ze stop listą i stemmingiem wspólny słownik 1-gramów ma 302 słowa (314 tylko ze stop listą).

```python
# cały zbiór III (11314 dokumentów): słownik i rozmiar reprezentacji 1 z n
from sklearn.feature_extraction.text import CountVectorizer

tokeny_III = {
    "bez stop/stem": [tokeny(d) for d in newsgroups.data],
    "stop + stemming": [tokeny(d, True, True) for d in newsgroups.data],
}

konf_III = [
    ("1-gramy",                       "bez stop/stem",   (1, 1)),
    ("1+2-gramy",                     "bez stop/stem",   (1, 2)),
    ("1+2+3-gramy",                   "bez stop/stem",   (1, 3)),
    ("1-gramy + stop + stemming",     "stop + stemming", (1, 1)),
    ("1+2+3-gramy + stop + stemming", "stop + stemming", (1, 3)),
]

macierze_BoW = {}
wiersze = []
for nazwa, wariant, zakres in konf_III:
    wektoryzator = CountVectorizer(analyzer=lambda toks, z=zakres: ngramy(toks, z))
    X = wektoryzator.fit_transform(tokeny_III[wariant])
    macierze_BoW[nazwa] = X
    V, N = X.shape[1], int(X.sum())          # V - rozmiar słownika, N - liczba tokenów w całym zbiorze
    wiersze.append({
        "konfiguracja": nazwa,
        "| rozmiar słownika V": V,
        "| liczba tokenów N": N
    })
tabela_III_1zn = pd.DataFrame(wiersze)
print(tabela_III_1zn.to_string(index=False))
```

**Czy jest możliwe utworzenie słownika dla wszystkich dokumentów ze zbioru III?** Tak, jest to możliwe. Co przedstawia kod powyżej.

**Jaki będzie rozmiar takiego słownika i reprezentacji 1 z n dla całego zbioru III?** Wartości z ostatniej tabeli (V to rozmiar słownika, N to liczba tokenów w całym zbiorze):
- 1-gramy: V = 154 921 słów, N = 3,4 mln tokenów.
- 1+2-gramy: V = 1,23 mln.
- 1+2+3-gramy: V = 3,28 mln.


**Czy istnieją struktury danych, które umożliwią zmniejszenie rozmiaru reprezentacji 1 z n?** Tak. Macierz 1 z n ma w każdym wierszu tylko jedną jedynkę, więc prawie wszystko to zera:
- **macierz rzadka** (`scipy.sparse`, formaty CSR/COO) przechowuje tylko niezerowe elementy.
- **lista numerów słów** (jeden indeks na token). To najkrótszy zapis, bo w wierszu wystarczy numer kolumny z jedynką;
- **hashing trick** (`HashingVectorizer`) nie wymaga przechowywania słownika, a wymiar jest stały;
- **redukcja słownika**: stop lista, stemming, małe litery, odcięcie rzadkich tokenów.

```python
# Zadanie 6
# Wykonaj zadanie 5 dla reprezentacji BoW.

# osobny słownik dla każdego dokumentu
wiersze = []
for nazwa_konf, zakres, stop, stemming in konfiguracje:
    for nazwa_dok, tekst in dokumenty.items():
        toks = ngramy(tokeny(tekst, stop, stemming), zakres)
        wek = CountVectorizer(analyzer=lambda t: t)
        X = wek.fit_transform([toks])
        wiersze.append({
            "konfiguracja": nazwa_konf,
            "dokument": nazwa_dok,
            "rozmiar słownika": X.shape[1],
            "kształt BoW": X.shape,
            "największa liczność": int(X.max()),
        })
print(pd.DataFrame(wiersze).to_string(index=False))

# przykład BoW dla dokumentu I (Roosevelt)
wek = CountVectorizer(analyzer=lambda t: t)
X = wek.fit_transform([tokeny(roosevelt_sentence)])
# czytelniejsza wersja w postaci ramki danych Pandas
pd.DataFrame(X.toarray(), columns=wek.get_feature_names_out())
```

```python
# wspólny słownik dla wszystkich 3 dokumentów
wiersze = []
for nazwa_konf, zakres, stop, stemming in konfiguracje:
    lista = [ngramy(tokeny(t, stop, stemming), zakres) for t in dokumenty.values()]
    wek = CountVectorizer(analyzer=lambda t: t)
    X = wek.fit_transform(lista)
    wiersze.append({
        "konfiguracja": nazwa_konf,
        "wspólny słownik": X.shape[1],
        "kształt BoW": X.shape,
        "niezerowe elementy": X.nnz,
        "gęsta [kB]": round(X.shape[0] * X.shape[1] * 4 / 1024, 1),
        "rzadka CSR [kB]": round((X.data.nbytes + X.indices.nbytes + X.indptr.nbytes) / 1024, 1),
    })
print(pd.DataFrame(wiersze).to_string(index=False))

# najczęstsze elementy BoW wspólnego słownika (1-gramy, bez stop listy)
wek = CountVectorizer(analyzer=lambda t: t)
X = wek.fit_transform([ngramy(tokeny(t)) for t in dokumenty.values()])
suma = np.asarray(X.sum(axis=0)).ravel()
najczestsze = np.argsort(-suma)[:10]
print([(wek.get_feature_names_out()[i], int(suma[i])) for i in najczestsze])
```

**Odpowiedzi do Zadania 6 (BoW)**

BoW dla dokumentu to **jeden wektor** zliczeń słów (kolejność tokenów jest tracona), a nie macierz z wierszem na każdy token.

- **Tokenizator i stop lista:** te same co w Zadaniu 5. BoW dla dokumentu to **jeden wektor** zliczeń (zamiast macierzy z wierszem na każdy token), więc jest dużo mniejszy. Stop lista skraca wektor i usuwa najczęstsze słowa (`the`, `of`), więc zostają słowa niosące treść.
- **2-gramy, 3-gramy, stemming, stop lista:** słownik zmienia się tak samo jak w Zadaniu 5 (n-gramy go powiększają, stemming i stop lista zmniejszają).  N-gramy rzadko się powtarzają, więc wektor jest bardziej rzadki.
- **Wspólny słownik dla 3 dokumentów:** BoW ma kształt 3 x V, gdzie każdy wiersz to jeden dokument, więc dokumenty można ze sobą porównywać.
- **Cały zbiór III:** BoW jest możliwy do zbudowania, bo ma jeden wiersz na dokument, a nie na token. Jest dużo mniejszy niż 1 z n.
- **Struktury zmniejszające rozmiar:** macierze rzadkie (CSR), słowniki z niezerowymi zliczeniami, hashing trick oraz redukcja słownika (stop lista, stemming, odcięcie rzadkich tokenów).
